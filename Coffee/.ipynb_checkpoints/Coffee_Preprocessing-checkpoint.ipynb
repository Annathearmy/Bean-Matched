```python

# 2. Assign the messy column to a variable for easy access
brew_col = "brew"

# 3. List the keywords you want to extract
# We use lowercase to keep it simple
methods_to_extract = ['Pour over', 'Espresso', 'French press', 'Drip coffee', 'AeroPress', 'Pod/capsule','Instant coffee']

# 4. The Loop: This is where the work happens
for method in methods_to_extract:
    # Create a clean column name (e.g., 'brew_espresso')
    new_col_name = f"brew_{method.lower().replace(' ', '_')}"

    # Check if the method exists in the string (case=False ignores Capitalization)
    # .fillna(False) ensures we don't crash on empty cells
    df[new_col_name] = df[brew_col].str.contains(method, case=False, na=False).astype(int)

# 5. Verify the results
print(df[[brew_col, 'brew_espresso', 'brew_pour_over']].head())
```

```python
df.drop("brew",axis=1,inplace= True)
df.head()
```

```python
df.fillna("Other",inplace=True)
df["favorite"].unique()
```

```python
df['additions'].unique()
```

```python
# 1. Define the column name from the CSV
add_col = "additions"

# 2. Fill empty cells (people who drink it black often leave this blank)
df[add_col] = df[add_col].fillna("None/Black")

# 3. List the core additions we want to track
add = ['Milk', 'Sugar', 'Syrup', 'Sweetener', 'Creamer']

# 4. The Implementation Loop
for a in add:
    # Create a clean name like 'adds_milk'
    new_name = f"adds_{a.lower()}"

    # Check if the word exists in the messy text
    df[new_name] = df[add_col].str.contains(a, case=False, na=False).astype(int)

# 5. Check the result
print(df[[add_col, 'adds_milk', 'adds_sugar']].head())
```

```python
from sklearn.preprocessing import LabelEncoder
le= LabelEncoder()
df["age"]= le.fit_transform(df["age"])
df.head()
```

```python
df['most_willing'].unique()
```

```python
def map_willingness(value):
    value = str(value).lower()
    if '$0-$2' in value or '$2-$4' in value:
        return 1 # Budget/Traditional
    elif '$4-$6' in value or '$6-$8' in value:
        return 2 # Mid-range/Specialty
    else:
        return 3 # Premium/Experimental

# Apply the mapping to a new clean column
df['willingness_score'] = df['most_willing'].apply(map_willingness)
```

```python
# Convert the expertise column to numbers
# 'coerce' turns text like "Other" into NaN (empty values)
df['expertise'] = pd.to_numeric(df['expertise'], errors='coerce')
# 3. Fill the 'Other' (NaN) spots with that average
df['expertise'] = df['expertise'].fillna(5)
```

```python
df["expertise"].unique()
```

```python
df["gender"].unique()
```

```python
# 1. Define your mapping dictionary
# This handles the most common entries in the James Hoffmann dataset
gender_map = {
    'Male': 0,
    'Female': 1,
    'Non-binary': 2,
    'Prefer not to say': 2,
    'Other': 2
}

# 2. Apply the mapping
# .get() handles any weird text by defaulting it to 2 (Other)
df['gender'] = df['gender'].apply(lambda x: gender_map.get(x, 2))

# 3. Verify the result
print(df['gender'].unique())
# You should now see: array([0, 1, 2])
```

```python
df["cups"].unique()
```

```python
# 1. Define the dictionary with exact strings from your array
cups_map = {
    'less than 1': 0.5,
    '1': 1.0,
    '2': 2.0,
    '3': 3.0,
    '4': 4.0,
    'more than 4': 5.0,
    'other': 2.0
}

# 2. Clean the column first: convert to string, lowercase, and remove spaces
df['cups'] = df['cups'].astype(str).str.lower().str.strip()

# 3. Apply the mapping , fill all the null values with avg
df['cups'] = df['cups'].map(cups_map).fillna(2.0)

# 4. Check again
print(df['cups'].unique())
```

```python
df['favorite'].unique()
```

Milk-Heavy (Low Intensity): Drinks like Lattes or Mochas dilute the coffee. Users need Dark Roasts (Coffee C) to "punch through" the milk.

Water-Diluted (Medium Intensity): Drip or Americanos prioritize "sessionability." Users typically prefer Balanced Roasts (Coffee B) for easy drinking.

Concentrated (High Intensity): Espresso and Cortados focus on texture and body. Users seek Medium to Dark profiles with heavy mouthfeel.

Pure/Filter (High Clarity): Pourovers have zero dilution. These users have the highest acidity tolerance and prefer Light/Fruity Roasts (Coffee A or D).

```python
# 1. Define the mapping
fav_map = {
    'Blended drink (e.g. Frappuccino)': 1,
    'Mocha': 1,
    'Latte': 1,
    'Cappuccino': 1,
    'Regular drip coffee': 2,
    'Iced coffee': 2,
    'Americano': 2,
    'Cold brew': 2,
    'Other': 2, # Default to the middle
    'Espresso': 3,
    'Cortado': 3,
    'Pourover': 4
}

# 2. Apply the mapping
df['favorite'] = df['favorite'].map(fav_map).fillna(2)

print(df['favorite'].unique())
```

```python
df["caffeine"]= df["caffeine"].replace("Other","Full caffeine")
df["caffeine"].unique()
```

```python
from sklearn.preprocessing import OrdinalEncoder
# 1. Define the logical order (Low to High caffeine)
caffeine_order = ['Decaf', 'Half caff', 'Full caffeine']

encoder = OrdinalEncoder(categories=[caffeine_order])

# 3. Fit and Transform the column
# Note: We reshape to (-1, 1) because the encoder expects a 2D array
df['caffeine'] = encoder.fit_transform(df[['caffeine']].fillna('Full caffeine'))

# Verify: Decaf=0.0, Half=1.0, Full=2.0
print(df['caffeine'].unique())
```
